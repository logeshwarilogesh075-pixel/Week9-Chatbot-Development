
# Week 9: Chatbot Development Using Ollama

## Objective
- Build an interactive AI chatbot using a local Large Language Model.
- Use Ollama with Python.
- Maintain conversation history.
- Save chatbot responses and timestamps.
- Evaluate the chatbot using sample questions.

## Technologies Used
- Python
- Ollama
- Llama 3.2 1B
- Pandas
- Jupyter Notebook


```python

import ollama
import pandas as pd
from pathlib import Path
from datetime import datetime

MODEL = "llama3.2:1b"

print("Libraries imported successfully!")
print("Model:", MODEL)
```

```python

result = ollama.chat(
    model=MODEL,
    messages=[
        {
            "role": "user",
            "content": "Explain artificial intelligence in simple English."
        }
    ]
)

print(result["message"]["content"])
```

```python

# Load saved conversation history

BASE_DIR = Path.cwd().parent
HISTORY_FILE = BASE_DIR / "dataset" / "chat_history.csv"

if HISTORY_FILE.exists():
    history_df = pd.read_csv(HISTORY_FILE)

    print("Total conversations:", len(history_df))
    display(history_df.tail(5))
else:
    print("Chat history file not found:", HISTORY_FILE)
```

```python

# Load responses generated by the chatbot

RESPONSES_FILE = BASE_DIR / "dataset" / "responses.csv"

if RESPONSES_FILE.exists():
    responses_df = pd.read_csv(RESPONSES_FILE)

    print("Total questions tested:", len(responses_df))
    print("\nFirst five chatbot responses:")
    display(responses_df.head())

    print("\nMissing responses:")
    print(responses_df["response"].isna().sum())

    print("\nQuestions with errors:")
    print(
        responses_df[
            responses_df["response"].astype(str).str.startswith("ERROR:")
        ][["question", "response"]]
    )
else:
    print("responses.csv not found.")
    print("Run python generate_responses.py in the terminal first.")
```


## Observations

2. It responds to educational questions about programming and artificial intelligence.
3. Conversation history is maintained during a chat session.
4. Chat conversations are saved with timestamps in a CSV file.
5. Sample questions help evaluate the chatbot's responses.
